# DuckDB and file formats

This tutorial is slightly based on the DuckDB [tutorial](https://duckdb.org/2024/05/31/analyzing-railway-traffic-in-the-netherlands.html).

Inputs: full 2024 [train services](https://www.rijdendetreinen.nl/en/open-data/train-archive) and 2023–2024 [disruptions](https://www.rijdendetreinen.nl/en/open-data/disruptions), provided by Rijden de Treinen (CC BY 4.0). Download/import commands are in `README.md`.

Each DuckDB connection is limited to 512 MB and two threads; temporary data spills under `data/duckdb_tmp/` on disk. The full dataset is retained.

## Reading from files

The only thing needed to run DuckDB is importing it. To compare with Postgres, we will read the CSV file with train services in 2024.

By the way, it's highly recommended to always use multi-line strings in triple quotes for all queries, even the simplest ones. It makes them more readable, easier to modify, and removes escaping quotes.

```python
import duckdb
import matplotlib.pyplot as plt

MEMORY_LIMIT = "512MB"
THREADS = 2
BLUE = "#466fa8"
ORANGE = "#b98236"
GRAY = "#a2a8b0"

duckdb.sql(f"""
SET memory_limit = '{MEMORY_LIMIT}';
SET threads = {THREADS};
SET temp_directory = 'data/duckdb_tmp/default';
SET preserve_insertion_order = false;
""")
plt.rcParams.update({"figure.figsize": (10, 4), "figure.constrained_layout.use": True})

duckdb.sql("""
SELECT *
FROM "data/services-2024.csv"
""")
```

Now that was fast and easy! DuckDB parses CSV and infers type with [quite advanced features](https://duckdb.org/docs/current/data/csv/auto_detection.html). You can also specify options manually ([documentation](https://duckdb.org/docs/current/data/csv/overview.html)), e.g.:

`FROM read_csv("flights.csv", delim = "|")`.

Jupyter Notebook automatically prints the results, but we could also save them and call `.show()` explicitly. It is returned as `DuckDBPyRelation` object ([documentation](https://duckdb.org/docs/current/clients/python/reference/#duckdb.DuckDBPyRelation)). You can use the object-oriented Relational API ([documentation](https://duckdb.org/docs/current/clients/python/relational_api.html)) instead of writing SQL to compose queries based on Python language constructs. It's less popular than SQL though, and can be less readable.

```python
results = duckdb.sql("""
SELECT *
FROM "data/services-2024.csv"
""")
print(type(results))
results.show()
```

```python
# last 10 rows
duckdb.sql("""
SELECT *
FROM "data/services-2024.csv"
ORDER BY "Service:Date"
LIMIT 10
""").show()
```

```python
# total number of late trains
duckdb.sql("""
SELECT COUNT(*)
FROM "data/services-2024.csv"
WHERE "Stop:Arrival delay" > 0
""")
```

Queries like above are transient (in-memory). They can temporarily spill data to disk to process larger-than-memory datasets, but don't save results. In particular, each query needs to parse CSV file.

To make results persistent, we can create `.db` (or `.duckdb`) file, similarly to SQLite. This is useful for a few reasons:
- you want to save work and go home
- for storing intermediate computation results for longer workflows
- avoiding reading and parsing files, or downloading remote data

Such file holds regular tables, which can be created from other data sources.

```python
db = duckdb.connect(
    "data/duckdb_trains.db",
    config={
        "memory_limit": MEMORY_LIMIT,
        "threads": THREADS,
        "temp_directory": "data/duckdb_tmp/persistent",
        "preserve_insertion_order": False,
    },
)
db
```

If you call methods on `db` object, they will be run on the database. If you use `duckdb` module, the default in-memory session will be used instead.

Let's create table from CSV and compare sizes with on-disk file.

```python
import os

db.sql("""
CREATE TABLE IF NOT EXISTS services AS
FROM "data/services-2024.csv"
""")

db.execute("CHECKPOINT")
csv_size_mb = os.path.getsize("data/services-2024.csv") // (1024 * 1024)
duckdb_size_mb = os.path.getsize("data/duckdb_trains.db") // (1024 * 1024)

print(f"CSV size: {csv_size_mb} MiB")
print(f"DuckDB size: {duckdb_size_mb} MiB")
```

DuckDB uses many ["friendly SQL"](https://duckdb.org/docs/current/sql/dialect/friendly_sql.html) extensions, which make SQL queries easier to write. They also offer many built-in features and functions. Some of them are DuckDB-exclusive, while others (e.g. PIVOT, UNPIVOT) are commonly adopted by many SQL databases.

Above, we use [FROM-first syntax](https://duckdb.org/docs/current/sql/query_syntax/from.html#from-first-syntax), which allows omitting `SELECT *`. Without it, the query would be `AS SELECT * FROM ...`.

It also allows switching order of SELECT and FROM. Many people find this more readable. Let's see this in action.

```python
db.sql("FROM services SELECT COUNT(*)")
```

Another friendly SQL extension is [DESCRIBE](https://duckdb.org/docs/current/guides/meta/describe.html), which provides a summary of table columns, data types, and their basic features. It's very useful to verify the data after loading CSV or other unreliable formats. It even allows skipping SELECT and FROM altogether.

```python
db.sql("DESCRIBE services")
```

Let's see a few examples of queries on this table:

```python
# number of all the services
db.sql("""
SELECT COUNT(DISTINCT "Service:RDT-ID") 
FROM services
""")
```

```python
# number of stops per service
db.sql("""
SELECT "Service:RDT-ID",
       COUNT(*) AS num_stops
FROM services
GROUP BY "Service:RDT-ID"
ORDER BY num_stops DESC
""")
```

```python
# average arrival delay per service type
db.sql("""
SELECT "Service:Type",
       AVG("Stop:Arrival delay") AS avg_arrival_delay
FROM services
WHERE "Stop:Arrival delay" IS NOT NULL
GROUP BY "Service:Type"
ORDER BY avg_arrival_delay DESC
""")
```

The ranking below counts scheduled arrivals plus departures (non-NULL timestamps), including cancelled schedules. A stop with both counts twice.

```python
# Rank scheduled arrival/departure events, rather than counting absent endpoints.
busiest_stations = db.sql("""
SELECT "Stop:Station code" AS station_code,
       "Stop:Station name" AS station,
       COUNT("Stop:Arrival time") + COUNT("Stop:Departure time") AS arrivals_departures
FROM services
GROUP BY station_code, station
ORDER BY arrivals_departures DESC, station_code
LIMIT 10
""").df()
display(busiest_stations)

db.sql("""
CREATE OR REPLACE TABLE disruptions AS
FROM read_csv('data/disruptions-2023.csv')
""")
display(db.sql("DESCRIBE disruptions").df())

display(
    db.sql("""
SELECT statistical_cause_en AS cause, COUNT(*) AS disruptions
FROM disruptions
GROUP BY cause
ORDER BY disruptions DESC, cause
LIMIT 5
""").df()
)

longest_disruption = db.sql("""
SELECT *, duration_minutes / 1440.0 AS duration_days
FROM disruptions
ORDER BY duration_minutes DESC NULLS LAST, rdt_id
LIMIT 1
""").df()
display(
    longest_disruption[
        ["rdt_id", "statistical_cause_en", "start_time", "end_time", "duration_days"]
    ]
)
```

## Connecting to databases

DuckDB can connect to various data sources, e.g. Excel, databases, cloud files, data lakehouses, and others. It uses a system of [extensions](https://duckdb.org/docs/current/extensions/overview), which are loaded explicitly. This avoids bloating the base package and is very flexible. We will now install and load the Postgres extension, and query the database.

Make sure that PostgreSQL is up and running in Podman Compose.

```python
# install and import extension
db.sql("""INSTALL postgres; LOAD postgres;""")

# NEVER write password like this explicitly in real code!
conn_string = "host=localhost user=postgres password=postgres dbname=postgres"

# note that we create the alias for this Postgres database here
# this also needs to be in single quotes
db.sql(f"""
ATTACH IF NOT EXISTS '{conn_string}' AS postgres_db_read (TYPE postgres, READ_ONLY);
""")
```

```python
# command to list all tables available
db.sql("SHOW ALL TABLES")
```

You can securely store connections information in [DuckDB secrets](https://duckdb.org/docs/current/core_extensions/postgres#configuring-via-secrets). Here, we make a read-only connection, but DuckDB can also [write to Postgres](https://duckdb.org/docs/current/core_extensions/postgres#writing-data-to-postgresql). This way you can e.g. insert Parquet data into Postgres, which is not supported natively.

When a database is attached, the data is queried from it each time. It guarantees the newest data, but also results in networking overhead. You can [copy Postgres table](https://duckdb.org/docs/current/core_extensions/postgres#usage) into DuckDB to avoid this, but you have stale data this way.

We have two `services` tables now - to query the Postgres one, we prefix it with database alias. Compare querying Postgres and DuckDB tables.

```python
db.sql("""
SELECT COUNT(*)
FROM postgres_db_read.services
""")
```

```python
db.sql("""
SELECT COUNT(*)
FROM services
""")
```

You can run any queries this way, e.g. JOIN files and databases, insert and update data sources, and more. To remove a database, use DETACH command:

```python
db.sql("DETACH postgres_db_read")
```

## Writing data

DuckDB can not only insert data into databases, but also write regular files to disk, for example [JSON and JSON Lines](https://duckdb.org/docs/current/data/json/writing_json) and [Parquet](https://duckdb.org/docs/current/data/parquet/overview). It also makes it very convenient as a format conversion tool, e.g. as a first part in a data processing workflow.

DuckDB is very fast at reading CSV, and [gets faster over time](https://duckdb.org/2024/06/26/benchmarks-over-time.html#csv-reader). However, its unique feature is [very robust and advanced CSV parsing](https://duckdb.org/2023/10/27/csv-sniffer.html), based on paper ["Multiple hypothesis CSV parsing" T. Döhmen et al.](https://hannes.muehleisen.org/publications/ssdbm2017-muehleisen-csvs.pdf), and in case of errors it has [verbose error informations](https://duckdb.org/docs/current/data/csv/reading_faulty_csv_files), as well as option to omit faulty lines.

Let's read the CSV file with 2024 train services and change it to TSV (tab-separated values) file. Exporting files also uses COPY command, with format-specific options ([documentation](https://duckdb.org/docs/current/sql/statements/copy.html#csv-options)).

```python
duckdb.sql("""
COPY (SELECT * FROM "data/services-2024.csv") TO "data/services_2024_duckdb.tsv" (HEADER, DELIMITER "\t");
""")

with open("data/services_2024_duckdb.tsv") as file:
    i = 0
    for i, line in enumerate(file):
        i += 1
        print(line)
        if i == 5:
            break
```

1. Convert the 2024 services file into JSON, JSON Lines and Parquet files. Use ZSTD compression for Parquet. Use [DuckDB documentation](https://duckdb.org/docs/current/sql/statements/copy.html#format-specific-options).
2. Compare file size of each file format (including CSV) in MB.
3. Compare loading time of each file format (including CSV), e.g. to count how many trains arrived at Rotterdam Centraal station. Use `time` module. Report mean time of 3 repetitions.

Times below measure the same scheduled-arrivals count query, including parsing and schema inference, over three consecutive runs per format. They are local query timings with operating-system caching, not cold full-file loading benchmarks. MB means 1,000,000 bytes; export time is excluded.

```python
from pathlib import Path
from statistics import mean
from time import perf_counter

import pandas as pd

REPETITIONS = 3
formats = [
    ("CSV", "data/services-2024.csv", "read_csv", None),
    ("JSON", "data/services-2024.json", "read_json_auto", "FORMAT JSON, ARRAY true"),
    ("JSONL", "data/services-2024.jsonl", "read_json_auto", "FORMAT JSON, ARRAY false"),
    (
        "Parquet",
        "data/services-2024.parquet",
        "read_parquet",
        "FORMAT PARQUET, COMPRESSION ZSTD",
    ),
]
rotterdam_arrivals = db.sql("""
SELECT COUNT(*) FROM services
WHERE "Stop:Station code" = 'RTD' AND "Stop:Arrival time" IS NOT NULL
""").fetchone()[0]
comparison_rows = []
for name, filename, reader, options in formats:
    if options is not None:
        db.execute(f"COPY services TO ? ({options})", [filename])

    durations = []
    for _ in range(REPETITIONS):
        started = perf_counter()
        count = db.execute(
            f"""
        SELECT COUNT(*) FROM {reader}(?)
        WHERE "Stop:Station code" = 'RTD' AND "Stop:Arrival time" IS NOT NULL
        """,
            [filename],
        ).fetchone()[0]
        durations.append(perf_counter() - started)
        assert count == rotterdam_arrivals

    comparison_rows.append(
        {
            "format": name,
            "size_mb": Path(filename).stat().st_size / 1_000_000,
            "mean_seconds": mean(durations),
            "arrivals": count,
        }
    )

format_comparison = pd.DataFrame(comparison_rows)
display(format_comparison.round(3))
axes = format_comparison.plot.bar(
    x="format",
    y=["size_mb", "mean_seconds"],
    subplots=True,
    layout=(1, 2),
    figsize=(10, 4),
    legend=False,
    rot=0,
    color=[BLUE, ORANGE],
)
axes[0, 0].set(
    title="Full 2024 services: file size", xlabel="Format", ylabel="Size (MB)"
)
axes[0, 1].set(
    title="Rotterdam scheduled-arrivals query",
    xlabel="Format",
    ylabel="Mean time (seconds)",
);
```

## DuckDB interoperability

DuckDB is highly interoperable with Python and with data processing frameworks like Pandas and Polars. It can convert results to them, as well as query DataFrames with SQL. This has a few uses:
- switch to SQL whenever you want
- use DuckDB for fast and robust data reading, e.g. for CSV
- convert to Pandas for plotting the results
- pool data from various sources with DuckDB and then build complex workflows with DataFrames

Getting results in Python works similarly to `psycopg`, with methods `.fetchone()` and `.fetchall()`.

Let's see how many services were run in January and December 2024.

```python
num_jan_services = duckdb.sql("""
SELECT COUNT(DISTINCT "Service:RDT-ID")
FROM "data/services-2024.csv"
WHERE "Service:Date" BETWEEN '2024-01-01' AND '2024-01-31'
""")
num_jan_services = num_jan_services.fetchone()[0]

num_dec_services = duckdb.sql("""
SELECT COUNT(DISTINCT "Service:RDT-ID")
FROM "data/services-2024.csv"
WHERE "Service:Date" BETWEEN '2024-12-01' AND '2024-12-31'
""")
num_dec_services = num_dec_services.fetchone()[0]

print(f"January services: {num_jan_services}")
print(f"December services: {num_dec_services}")
```

Converting to Pandas is similar, with method `.to_df()` (aliases: `.fetch_df()`, `.df()`). Let's fetch all services by month.

```python
num_monthly_services = duckdb.sql("""
SELECT MONTH("Service:Date") AS month, COUNT(DISTINCT "Service:RDT-ID") AS num_services 
FROM "data/services-2024.csv"
GROUP BY MONTH("Service:Date")
ORDER BY month
""").df()

num_monthly_services
```

Since we're already here, let's visualize this with Matplotlib. Pandas has handy shortcuts in `.plot` method ([documentation](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.plot.html)).

```python
num_monthly_services.plot.line(
    x="month",
    y="num_services",
    title="Distinct services in 2024 by month",
    xlabel="Month",
    ylabel="Distinct services",
    legend=False,
    color=BLUE,
);
```

We can also query Pandas DataFrames by just using variable names like tables. Note that we should use Arrow backend for good performance here. Let's load the disruptions from 2023 and check for frequency of top 10 most popular causes. Then we will go back to Pandas and make a bar plot.

```python
from matplotlib.ticker import PercentFormatter

import pandas as pd

df_disruptions = pd.read_csv("data/disruptions-2023.csv", dtype_backend="pyarrow")

df_disruptions_statistics = duckdb.sql("""
SELECT statistical_cause_en AS cause, COUNT(*) / (SELECT COUNT(*) FROM df_disruptions) AS cause_frequency
FROM df_disruptions
GROUP BY statistical_cause_en
ORDER BY cause_frequency DESC
LIMIT 10
""").df()

ax = df_disruptions_statistics.plot.barh(
    x="cause",
    y="cause_frequency",
    title="Most frequent disruption causes in 2023",
    xlabel="Share of all 2023 disruptions",
    ylabel="Cause",
    legend=False,
    figsize=(10, 6),
    color=BLUE,
)
ax.invert_yaxis()
ax.xaxis.set_major_formatter(PercentFormatter(1))
```

1. Extract the average daily arrival delays of 2024 train services with DuckDB. Note that NULL delay means zero, and you have to use [COALESCE function](https://duckdb.org/docs/current/sql/functions/utility.html#coalesceexpr-) to fill them as zeros.

This lab-defined mean includes every stop row and replaces missing delays with zero, including origins without scheduled arrivals. It is not a mean restricted to observed arrivals. Rolling means use complete trailing 7- and 30-day windows.

```python
daily_delays = (
    db.sql("""
SELECT "Service:Date" AS date,
       AVG(COALESCE("Stop:Arrival delay", 0)) AS "Daily mean"
FROM services
GROUP BY date
ORDER BY date
""")
    .df()
    .set_index("date")
    .asfreq("D")
)
daily_delays["7-day mean"] = daily_delays["Daily mean"].rolling(7).mean()
daily_delays["30-day mean"] = daily_delays["Daily mean"].rolling(30).mean()
daily_delays.plot.line(
    title="2024 arrival delay: daily and trailing averages",
    xlabel="Service date",
    ylabel="Mean delay (minutes)",
    color=[GRAY, BLUE, ORANGE],
    style=["-", "-", "--"],
);
```
